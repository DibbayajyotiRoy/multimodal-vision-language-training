# VLM Error Analysis

This notebook loads the analysis report produced by `src/analysis.py` and visualises outliers, false positives, and data-quality issues.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

REPORT_PATH = Path("outputs/analysis/error_report.json")
report = json.loads(REPORT_PATH.read_text())

print("Outliers:", len(report["outliers"]))
print("Model problems:", report["false_positive_report"]["model_problems_count"])
print("Data problems:", report["false_positive_report"]["data_problems_count"])
```

```python
losses = [r["loss"] for r in report["outliers"]]
plt.figure(figsize=(10, 4))
plt.hist(losses, bins=20)
plt.xlabel("Loss")
plt.ylabel("Count")
plt.title("Distribution of Outlier Losses")
plt.show()
```

```python
for problem in report["false_positive_report"]["model_problems"][:5]:
    print(f"Image: {problem['image_id']}")
    print(f"Prediction: {problem['prediction']}")
    print(f"Reference: {problem['reference']}")
    print(f"Issue: {problem['issue']}")
    print("-" * 40)
```
